# Cria DF Treinamento

```python
def balance_classes(df):
    
    n3 = len(df[df['SENTIMENT'] == 'POSITIVO'])
    
    n2 = len(df[df['SENTIMENT'] == 'NEUTRO'])
    
    n1 = len(df[df['SENTIMENT'] == 'NEGATIVO'])
    
    aux_min = min(n1,n2,n3)
    #
    df_positivo = df.loc[df['SENTIMENT'] == 'POSITIVO', :]

    #
    df_negativo = df.loc[df['SENTIMENT'] == 'NEGATIVO', :]
    
    df_neutro = df.loc[df['SENTIMENT'] == 'NEUTRO', :]
   
    #
    df_positivo = df_positivo.sample(aux_min)
    
    df_negativo = df_negativo.sample(aux_min)
    
    df_neutro = df_neutro.sample(aux_min)
    
    # constrói a tabela final
    df_final = df_positivo.append(df_negativo).reset_index().drop(columns=['index'])
    
    df_final = df_final.append(df_neutro).reset_index().drop(columns=['index'])
    
    return df_final


#df_final =  inputs[0]
#df_final = df_final.append(inputs[1])
#df_final = df_final.append(inputs[2])
#df_final = balance_classes(df_final)

df_total = df1.append(df2)

df_total = df_total.append(df3)

df_balance = balance_classes(df_total)
```
