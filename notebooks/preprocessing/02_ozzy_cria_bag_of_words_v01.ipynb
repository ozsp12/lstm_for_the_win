# Pipeline Bag of Words

```python
import class_nlp_preprocess as nlppreprocess

obj_nlprepro = nlppreprocess.class_train_model()
```

```python
# Cria o dataframe inicial com os comentários raspados
df = obj_nlprepro.read_df_file()

# Faz o preprocessamento da base para iniciar vetorização e tokenização
df_final = obj_nlprepro.preprocess_data(df, 
                                        columns = ['COMMENT_TEXT', 'RATING_VALUE'],
                                        label_column='RATING_VALUE', 
                                        null = True,
                                        labels_ok = False,
                                        dict_labels = {(' '): '', (','): '.'},
                                        labels_ready = False, 
                                        labels_true = {(5,5): 2, (4,4):0,(1,2,3):1}
                                       )

# contruindo a coluna nova com os sentimentos
df_final['sentimentos'] = df_final['RATING_VALUE'].apply(obj_nlprepro.sentimento)

# Faz o balanço das classes
df_final = obj_nlprepro.balance_classes(df_final)

# tokenização
df_final = obj_nlprepro.tokenizacao(df_final)

# cria a lista de treino com as palavras para o modelo
lista_treino = obj_nlprepro.treinar_vetorizacao(df_final)

# criando a coluna com o texto vetorizado
df_final['vectors'] = df_final['token'].apply(obj_nlprepro.vectorize2)

# cria o dicionário para o bag of words
df_final['dic_bag'] = df_final['token'].apply(obj_nlprepro.dic_bag_of_worlds)

# cria a bag of words
df_dic_bag = obj_nlprepro.df_dic_bag(df_final,'token')
```
